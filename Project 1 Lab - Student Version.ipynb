```python
import pandas as pd
```

```python
# Load the CSV file 

air_data = pd.read_csv("AirQuality_Daily_StudentVersion.csv")

# Turn data into a dataframe
air_data = pd.DataFrame(air_data)


# Preview first 5 rows to confirm the data loaded correctly
air_data.head(5)
```

```python
# Show all column names to see the titles of each columns
air_data.columns
```

```python
# Group data by sensor.name and count number of observations per sensor
sensor_counts = air_data.groupby(["sensor.name"]).size()

# Print the counts
print(sensor_counts)
```

```python
# Group by sensor.name and compute mean and median for VOC, PM2.5, and PM10
grouped_data = air_data.groupby("sensor.name").agg(
    voc_mean=("voc", "mean"),
    voc_median=("voc", "median"),
    pm25_mean=("pm2.5_atm", "mean"),
    pm25_median=("pm2.5_atm", "median"),
    pm10_mean=("pm10.0_atm", "mean"),
    pm10_median=("pm10.0_atm", "median")
)

# Print the summary table
print(grouped_data.head())
```

```python
# Top 5 sensors by VOC mean
top5_voc_mean = grouped_data.sort_values(by="voc_mean", ascending=False).head(5)
print(top5_voc_mean)

# Top 5 sensors by VOC median
top5_voc_median = grouped_data.sort_values(by="voc_median", ascending=False).head(5)
print(top5_voc_median)


# Top 5 sensors by PM2.5 mean
top5_pm25_mean = grouped_data.sort_values(by="pm25_mean", ascending=False).head(5)
print(top5_pm25_mean)

# Top 5 sensors by PM2.5 median
top5_pm25_median = grouped_data.sort_values(by="pm25_median", ascending=False).head(5)
print(top5_pm25_median)

# Top 5 sensors by PM10 mean
top5_pm10_mean = grouped_data.sort_values(by="pm10_mean", ascending=False).head(5)
print(top5_pm10_mean)

# Top 5 sensors by PM10 median
top5_pm10_median = grouped_data.sort_values(by="pm10_median", ascending=False).head(5)
print(top5_pm10_median)
```

```python
# Find the maximum VOC value.
highest_voc = air_data["voc"].max()

# find the date and sensor of the maximum VOC
max_voc_event = air_data[air_data["voc"] == highest_voc][
    ["date", "sensor.name", "voc"]
]

print(max_voc_event)

```

```python
#now we find the maximum pm2.5 value

highest_pm25 = air_data["pm2.5_atm"].max()

#we find the date and sensor of the maximum pm2.5
air_data[air_data["pm2.5_atm"] == highest_pm25][
    ["date", "sensor.name", "pm2.5_atm"]
]
```

```python
#now we find the maximum pm10.0 value

highest_pm10 = air_data["pm10.0_atm"].max()

#we find the date and sensor of the maximum pm10.0

air_data[air_data["pm10.0_atm"] == highest_pm10][
    ["date", "sensor.name", "pm10.0_atm"]
]
```

```python

#define a function that assigns the humidity levels

def humidity_category(humidity):
    if humidity < 50:
        return "Low"
    elif 50 <= humidity <= 80:
        return "High"
    else:
        return "Very High"

#Apply altitude categories to the full dataset
air_data["humidity_category"] = air_data["humidity"].apply(humidity_category)

air_data.head(30)


```

```python
# Define temperature category function
def temperature_category(x):
    if x < 32:
        return "Below Freezing"
    elif 32 <= x <= 50:
        return "Cool"
    elif 51 <= x <= 70:
        return "Warm"
    else:
        return "Hot"


#Apply temperature categories to the full dataset
air_data["temperature_category"] = air_data["temperature"].apply(temperature_category)

air_data.head(10)

        

        
```

```python
# Group by humidity and temperature categories and calculate mean pollutant values
weather_grouped = air_data.groupby(["humidity_category", "temperature_category"]).agg(
    voc_mean=("voc", "mean"),
    pm25_mean=("pm2.5_atm", "mean"),
    pm10_mean=("pm10.0_atm", "mean")
)

print(weather_grouped)
```

```python
# Filter PM2.5 values in the AQI "Unhealthy for Sensitive Groups" range
pm25_usg = air_data[
    (air_data["pm2.5_atm"] >= 35.5) &
    (air_data["pm2.5_atm"] <= 55.4)
]

#Select only date, sensor, and PM2.5 columns and sort by date
pm25_usg_events = pm25_usg[
    ["date", "sensor.name", "pm2.5_atm"]
].sort_values("date")

#Display first 10 events
pm25_usg_events.head(10)
```

```python
# Filter PM10 values in the AQI "Unhealthy for Sensitive Groups" range
pm10_usg = air_data[
    (air_data["pm10.0_atm"] >= 155) &
    (air_data["pm10.0_atm"] <= 254)
]

#Select only date, sensor, and PM10 columns and sort by date
pm10_usg_events = pm10_usg[
    ["date", "sensor.name", "pm10.0_atm"]
].sort_values("date")

#Display first 10 events
pm10_usg_events.head(10)
```

```python
# Step 1: filter data that has VOC > 25
unhealthy_voc = air_data.loc[air_data['voc'] > 25]

# Step 2: keep only needed columns and sort by date
unhealthy_events = (
    unhealthy_voc[['date', 'sensor.name', 'voc']]
    .sort_values(by='date')
)

unhealthy_events.head(10)
```

```python
# Group air quality data by sensor name and sensor altitude
altitude_air_quality = air_data.groupby(
    ["sensor.name", "sensor.altitude"]
).agg(
    voc_mean=("voc", "mean"),
    pm25_mean=("pm2.5_atm", "mean"),
    pm10_mean=("pm10.0_atm", "mean")
)

# Display summarized air quality values by sensor and altitude
print(altitude_air_quality)
```
